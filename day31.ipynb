```python
def pali_slicer(s):
    n = len(s)
    isPalin = [[False]*n for _ in range(n)]

    for i in range(n-1, -1, -1):       # i goes backward
        for j in range(i, n):           # j goes forward from i
            if i == j:
                isPalin[i][j] = True
            elif j == i+1:
                isPalin[i][j] = (s[i] == s[j])
            else:
                isPalin[i][j] = (s[i] == s[j]) and isPalin[i+1][j-1]
    dp = [0] * n
    for i in range(n):
        if isPalin[0][i]:           # whole prefix is already a palindrome
            dp[i] = 0
        else:
            dp[i] = float('inf')
            for j in range(1, i+1):
                if isPalin[j][i]:    # piece from j to i is a palindrome
                    dp[i] = min(dp[i], dp[j-1] + 1)
    return dp[n-1]
s = "ab"
pali_slicer(s)
```
