```python
def fibonnaci(n):
    if n <= 1:
        return 1
    dp = [0]*(n+1)
    dp[0] = 0
    dp[1] = 1
    for i in range(2,n+1):
        dp[i] = dp[i-1] + dp[i-2]
    return dp[n]
fibonnaci(100)
```

```python
'''
Climbing Stairs.
You're climbing a staircase with n steps. Each time you can climb 1 or 2 steps. How many distinct ways can you reach the top?
n = 2 → answer = 2    # (1+1) or (2)
n = 3 → answer = 3    # (1+1+1) or (1+2) or (2+1)
n = 5 → answer = 8
'''
def climbingStairs(n):
    if n <= 1:
        return 1
    dp = [0]*(n+1)
    dp[0] = 1
    dp[1] = 2
    for i in range(2,n+1):
        dp[i] = dp[i-1] + dp[i-2]
    return dp[n]
climbingStairs(2)
```

```python
'''
House Robber.
You're a robber. Houses are in a line, each with some amount of money. You can't rob two adjacent houses. Find the maximum amount you can rob.
nums = [1, 2, 3, 1]  → answer = 4   # rob house 0 and house 2
nums = [2, 7, 9, 3, 1]  → answer = 12  # rob house 0, 2, 4
'''
def houseRobber(nums):
    dp = [0]*len(nums)
    dp[0],dp[1]=nums[0],max(nums[0],nums[1])

    for i in range(2,len(nums)):
        dp[i] = max(dp[i-1] , nums[i] + dp[i-2])
    
    return dp[len(nums)-1]

nums = [1, 2, 0, 1,7]
print(houseRobber(nums))
```

```python
def coinChange(coins,amount):
    dp = [float('inf')] * (amount+1)
    dp[0] = 0

    for i in range(1, amount + 1):      # every amount
        for coin in coins:               # try every coin
            if coin <= i:
                dp[i] = min(dp[i], dp[i-coin] + 1)
        
    return dp[amount] if dp[amount] != float('inf') else -1

coins = [1,3,4]
amount = 6

print(coinChange(coins,amount))
```
