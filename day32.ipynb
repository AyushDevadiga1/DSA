```python
'''
Two Sum — LeetCode #1
Given an array of integers nums and an integer target, return the indices of the two numbers that add up to target.

You may assume each input has exactly one solution. You may not use the same element twice.
nums = [2, 7, 11, 15] target = 9 output: [0, 1]
nums = [3, 2, 4] target = 6 output: [1, 2]
'''

def two_sum(nums,target):
    hash = {}
    for i,val in enumerate(nums):
        diff = target - val 
        if diff in hash:
            return [hash[diff],i]
        hash[val] = i
nums = [3, 2, 4] 
target = 6
two_sum(nums,target)
```

```python
'''
**Problem:** Given an array of integers `nums` and an integer `k`, return the total number of subarrays whose elements sum to exactly `k`.
**Examples:**
nums = [1, 2, 3], k = 3
answer = 2        # [1,2] and [3]

nums = [1, 1, 1], k = 2
answer = 2        # [1,1] at index 0-1, and [1,1] at index 1-2

nums = [1, -1, 1], k = 1
answer = 3        # [1], [1,-1,1], [1]

nums = [3], k = 3
answer = 1        # [3] single element

nums = [1, 2, 4], k = 3
answer = 1        # only [1,2]
'''

def subarrays(nums,k):
    counter = 0
    prefix = 0
    hash = {0:1}
    for num in nums:
        prefix+=num
        counter = hash.get(prefix-k,0)
        hash[prefix] =  hash.get(prefix,0)+1
    return counter

nums = [1,-1,1]
k = 1
subarrays(nums,k)
```
