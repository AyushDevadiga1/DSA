````python
'''
Two Sum — LeetCode #1
Given an array of integers nums and an integer target, return the indices of the two numbers that add up to target.

You may assume each input has exactly one solution. You may not use the same element twice.
nums = [2, 7, 11, 15] target = 9 output: [0, 1]
nums = [3, 2, 4] target = 6 output: [1, 2]
'''

def two_sum(nums,target):
    hash = {}
    for i,val in enumerate(nums):
        diff = target - val 
        if diff in hash:
            return [hash[diff],i]
        hash[val] = i
nums = [3, 2, 4] 
target = 6
two_sum(nums,target)
````

````python
'''
**Problem:** Given an array of integers `nums` and an integer `k`, return the total number of subarrays whose elements sum to exactly `k`.
**Examples:**
nums = [1, 2, 3], k = 3
answer = 2        # [1,2] and [3]

nums = [1, 1, 1], k = 2
answer = 2        # [1,1] at index 0-1, and [1,1] at index 1-2

nums = [1, -1, 1], k = 1
answer = 3        # [1], [1,-1,1], [1]

nums = [3], k = 3
answer = 1        # [3] single element

nums = [1, 2, 4], k = 3
answer = 1        # only [1,2]
'''

def subarrays(nums,k):
    counter = 0
    prefix = 0
    hash = {0:1}
    for num in nums:
        prefix+=num
        counter = hash.get(prefix-k,0)
        hash[prefix] =  hash.get(prefix,0)+1
    return counter

nums = [1,-1,1]
k = 1
subarrays(nums,k)
````

````python
'''

**3Sum** — same family, one level up.

Given an array `nums`, find all unique triplets that sum to zero.

nums = [-1, 0, 1, 2, -1, -4]
answer = [[-1, -1, 2], [-1, 0, 1]]

nums = [0, 0, 0]
answer = [[0, 0, 0]]

nums = [1, 2, 3]
answer = []

'''

def three_sum(nums,target):
    answer = []
    nums.sort()
    for i in range(len(nums)):
        if i>0 and nums[i] == nums[i-1]:
            continue
        j,k = i+1,len(nums)-1
        while(j<k):
            current_sum = nums[i]+nums[j]+nums[k]
            if current_sum == target:
                answer.append([nums[i],nums[j],nums[k]])
                j+=1
                k-=1
            elif current_sum < k:
                j+=1
            else:
                k-=1
    return answer
nums = [-1, 0, 1, 2, -1, -4]
three_sum(nums,target=0)
````

````python
'''
**4Sum** — same family, one more level.

Given an array `nums` and a target `k`, find all unique quadruplets that sum to `k`.

```
nums = [1, 0, -1, 0, -2, 2], k = 0
answer = [[-2, -1, 1, 2], [-2, 0, 0, 2], [-1, 0, 0, 1]]

nums = [2, 2, 2, 2, 2], k = 8
answer = [[2, 2, 2, 2]]

'''
def four_sum(nums,target):
    answer = []
    nums.sort()
    for i in range(len(nums)):
        if i>0 and nums[i] == nums[i-1]:
            continue
        for j in range(i+1,len(nums)):
            if j>i+1 and nums[j] == nums[j-1]:
                continue
            k,l = j+1,len(nums)-1
            while(k<l):
                current_sum = nums[i]+nums[j]+nums[k]+nums[l]
                if current_sum == target:
                    answer.append([nums[i],nums[j],nums[k],nums[l]])
                    k+=1
                    l-=1
                    while k < l and nums[k] == nums[k-1]:
                        k += 1
                    while k < l and nums[l] == nums[l+1]:
                        l -= 1
                elif current_sum < target:
                    k+=1
                else:
                    l-=1
    return answer
nums = [1, 0, -1, 0, -2, 2]
target = 0
four_sum(nums,target)
````
