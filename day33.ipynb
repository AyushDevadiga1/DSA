```python
'''
Given a binary tree, determine if it is height-balanced. A tree is balanced if for every node, the height difference between left and right subtrees is at most 1.
      3
     / \
    9  20
       / \
      15   7
answer = True

      1
     / \
    2   2
   / \
  3   3
 / \
4   4
answer = False
'''

diameter = 0

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
    

def isBalanced(node):
    
    if node is None:
        return 0
    
    left = isBalanced(node.left)
    right = isBalanced(node.right)

    if left == -1 or right == -1:
        return -1

    if abs(left-right)>1:
        return -1
    
    return max(left,right)+1
    

root = TreeNode(3)
root.left = TreeNode(9)
root.right = TreeNode(20)
root.right.left = TreeNode(15)
root.right.left.left = TreeNode(15)
print(isBalanced(root) != -1)
```

```python
''' 
Problem 6:
Given the head of a linked list, return the node where the cycle begins. If there is no cycle return None.
1 → 2 → 3 → 4 → 5 → 3 (cycle back to node with value 3)
answer = node with value 3

1 → 2 → 3
answer = None
'''

class Node:
    def __init__(self,val):
        self.val = val
        self.next = None
        
def detectCycle(head):
    seen = set()
    node = head
    while node.next is not None:
        if node.val in seen:
            return node.val
        seen.add(node.val)
        node = node.next
    return None
node1 = Node(1)
node2 = Node(2)
node3 = Node(3)
node4 = Node(4)
node5 = Node(5)
node1.next = node2
node2.next = node3
node3.next = node4
node4.next = node5
node5.next = node3 

detectCycle(node1)
```

```python
''' 
Problem 6:

Given a binary tree and a target sum, return True if there exists a root-to-leaf path that sums to target.
"""      5
     / \
    4   8
   /   / \
  11  13   4
 /  \       \
7    2        1"""

target = 22 → True   # path 5→4→11→2
target = 100 → False
'''

class TreeNode:
    def __init__(self,val,left=None,right=None):
        self.val = val
        self.left = left
        self.right = right
    
def pathSum(node, target):
    if node is None:
        return False
    target-=node.val
    if target == 0 and node.left is None and node.right is None:
        return True
    return pathSum(node.left,target) or pathSum(node.right,target)

root = TreeNode(5)

root.left = TreeNode(4)
root.left.left = TreeNode(11)
root.left.left.left = TreeNode(7)
root.left.left.right = TreeNode(2)

root.right = TreeNode(8)
root.right.left = TreeNode(13)
root.right.right = TreeNode(4)
root.right.right.right = TreeNode(1)

target = 22
pathSum(root,target=target)
```

```python
'''
Longest Substring with At Most K Distinct Characters
Given a string s and integer k, find the length of the longest substring that contains at most k distinct characters.
s = "eceba", k = 2
answer = 3        # "ece"

s = "aabbcc", k = 1
answer = 2        # "aa" or "bb" or "cc"

s = "aabbcc", k = 2
answer = 4        # "aabb" or "bbcc"

s = "abc", k = 3
answer = 3        # whole string
'''

def max_k(s,k):
    seen = {}
    i=0
    answer = 0
    for j in range(len(s)):
        # add current character to map, store its count
        seen[s[j]] = seen.get(s[j], 0) + 1
        
        # shrink from left when distinct chars exceed k
        while len(seen) > k:
            seen[s[i]] -= 1
            if seen[s[i]] == 0:
                del seen[s[i]]
            i += 1
        
        answer = max(answer, j - i + 1)
    return answer
s = "eceba"
k = 2
max_k(s,k)
```

```python
'''
Problem 4:
Given a string, find the first non-repeating character and return its index. If none return -1.
s = "leetcode"   → 0
s = "loveleetcode" → 2
s = "aabb" → -1
'''
def first_non_repeating_char(s):
    if len(s) == 1:
        return 0
    from collections import Counter
    hash = Counter(s)
    for i,j in hash.items():
        if j == 1:
            for l,k in enumerate(s):
                if k == i:
                    return l
    return -1

s = "loveleetcode"
first_non_repeating_char(s)
```
