```python
'''
Given an array of positive integers nums and a target k, find the length of the shortest subarray whose sum is greater than or equal to k.
nums = [2, 3, 1, 2, 4, 3], k = 7
answer = 2        # [4, 3]

nums = [1, 1, 1, 1, 1], k = 7
answer = 0        # impossible, total sum is only 5

nums = [1, 4, 3, 2], k = 6
answer = 2        # [4,3] or [3,2] or... wait, check yourself
'''

def shortest_subarray(nums, k):
    answer = float('inf')
    i=0
    prefix = 0
    for j in range(len(nums)):
        prefix+=nums[j]
        while(prefix>=k):
            prefix-=nums[i]
            answer = min(answer,j-i+1)
            i+=1
    return answer
nums = [2, 3, 1, 2, 4, 3]
k = 7
print(shortest_subarray(nums, k))  # Output: 2 ([4, 3])
```

```python
'''
Longest Substring Without Repeating Characters
Given a string, find the length of the longest substring without any repeating characters.
s = "abcabcbb"
answer = 3        # "abc"

s = "bbbbb"
answer = 1        # "b"

s = "pwwkew"
answer = 3        # "wke"

s = "abcd"
answer = 4        # whole string
'''

def longest_substring(s):
    seen = {}
    i = 0
    answer = 0
    for j in range(len(s)):
        if s[j] in seen and seen[s[j]]>=i:
            i = seen[s[j]] + 1
        seen[s[j]] = j
        answer = max(answer,j-i+1)
    return answer
s = "abcabcbb"
longest_substring(s)
```

```python
'''
Longest Substring with At Most K Distinct Characters
Given a string s and integer k, find the length of the longest substring that contains at most k distinct characters.
s = "eceba", k = 2
answer = 3        # "ece"

s = "aabbcc", k = 1
answer = 2        # "aa" or "bb" or "cc"

s = "aabbcc", k = 2
answer = 4        # "aabb" or "bbcc"

s = "abc", k = 3
answer = 3        # whole string
'''
def longest_substring_distinct(s,k):
    seen = {}
    i = 0
    answer = 0
    for j in range(len(s)):
        seen[s[j]] = seen.get(s[j],0)+1
        while len(seen)>k:     
            seen[s[i]]-=1       
            if seen[s[i]] == 0:
                del seen[s[i]]
            i+=1
        answer = max(answer,j-i+1)
    return answer

s = "eceba"
k = 2
longest_substring_distinct(s,k)
```

```python
'''
Binary Search — Find a target in a sorted array.
nums = [-1, 0, 3, 5, 9, 12], target = 9
answer = 4        # index of 9

nums = [-1, 0, 3, 5, 9, 12], target = 2
answer = -1       # not found
'''

def binary_search(nums,target):
    low,high = 0,len(nums)-1
    while low<=high:
        mid = (low+high)//2
        if nums[mid] == target:
            return mid
        elif nums[mid] > target:
            high = mid - 1
        else:
            low = mid + 1
    return -1

nums = [-1, 0, 3, 5, 9, 12]
target = 9
binary_search(nums,target)
```
