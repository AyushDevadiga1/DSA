```python
'''  
Problem 1 — 20 minutes, no hints unless stuck.
You are climbing a staircase with n steps. You can climb 1, 2, or 3 steps at a time. How many distinct ways can you reach the top?
n = 3 → answer = 4    # (1+1+1), (1+2), (2+1), (3)
n = 4 → answer = 7
'''
```

```python
def three_steps(n):
    dp = [1]*(n+1)
    dp[2] = 2
    for i in range(3,n+1):
        dp[i] = dp[i-1] + dp[i-2] + dp[i-3]
    return dp[n]
three_steps(3)
```

```python
''''  
Problem 2 — 20 minutes.
Given a triangle array, find the minimum path sum from top to bottom. At each step you can move to an adjacent number in the row below.
triangle = [
     [2],
    [3,4],
   [6,5,7],
  [4,1,8,3]
]
answer = 11   # 2 + 3 + 5 + 1
'''
```

```python
def triangle_min_path(triangle):
    n = len(triangle)
    dp = triangle[-1][:]   # copy the last row as starting point
    
    for i in range(n-2, -1, -1):       # go from second-last row up to row 0
        for j in range(len(triangle[i])):
            dp[j] = triangle[i][j] + min(dp[j], dp[j+1])
    
    return dp[0]

triangle = [
     [2],
    [3,4],
   [6,5,7],
  [4,1,8,3]
]
triangle_min_path(triangle)
```
