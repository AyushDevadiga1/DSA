```python
heights = [0,1,0,2,1,0,1,3,2,1,2,1]

def max_water(heights):
    answer = 0
    max_left = [0]*len(heights)
    max_right = [0]*len(heights)
    max_left[0] = heights[0]
    max_right[-1] = heights[-1]

    for i in range(1,len(heights)):
        max_left[i] = max(max_left[i-1],heights[i])

    for i in range(len(heights)-2,-1,-1):
        max_right[i] = max(max_right[i+1],heights[i])

    for i in range(len(heights)):
        answer += min(max_left[i],max_right[i]) - heights[i]

    return answer

max_water(heights)
```

```python
# Now using floyd's cycle algo also
# 1 → 2 → 3 → 4 → 5 → 3 (cycle back to node with value 3)

class Node:
    def __init__(self,val):
        self.val = val
        self.next = None
        
def detectCycle(head):

    visited = set()
    current = head

    while current is not None:
        if current in visited:
            return current.val
        visited.add(current)
        current = current.next
    return None

def floyd_cycle(head):
    slow,fast = head,head.next
    while slow.next is not None:
        if slow.val == fast.val:
            return slow.val
        slow = slow.next
        fast = fast.next.next
    return -1

node1 = Node(1)
node2 = Node(2)
node3 = Node(3)
node4 = Node(4)
node5 = Node(5)
node1.next = node2
node2.next = node3
node3.next = node4
node4.next = node5
node5.next = node3 

print(detectCycle(node1))
print(floyd_cycle(node1))
```
