```python
'''
Problem 8:
Given n non-negative integers representing an elevation map where the width of each bar is 1, compute how much water it can trap after raining.
heights = [0,1,0,2,1,0,1,3,2,1,2,1]
answer = 6

heights = [4,2,0,3,2,5]
answer = 9
'''
```

```python
def water_trap(heights):
    water = 0

    max_left = [0] * len(heights)
    max_left[0] = heights[0]
    for i in range(1, len(heights)):
        max_left[i] = max(max_left[i-1], heights[i])

    max_right = [0] * len(heights)
    max_right[-1] = heights[-1]
    for i in range(len(heights)-2,-1,-1):
        max_right[i] = max(max_right[i+1], heights[i])

    for i in range (len(heights)) : 
        water += min(max_left[i], max_right[i]) - heights[i]
    return water

heights = [4,2,0,3,2,5]
water_trap(heights)
```

```python
'''
Problem 1:
Given a binary array nums, return the maximum number of consecutive 1s.
nums = [1,1,0,1,1,1]
answer = 3

nums = [1,0,1,1,0,1]
answer = 2
'''
```

```python
def consecutive_1(nums):
    max_1 = 0
    count = 0
    for num in nums:
        if num == 1:
            count+=1
        else:
            max_1 = max(max_1,count)
            count = 0
    max_1 = max(max_1,count)
    return max_1

nums = [1,0,1,1,0,1]
consecutive_1(nums)
```

```python
'''
Problem 4:
Given a string, find the first non-repeating character and return its index. If none return -1.
s = "leetcode"   → 0
s = "loveleetcode" → 2
s = "aabb" → -1
'''
```

```python
def first_non_repeating_char(s):
    if s[0] != s[1]:
        return 0
    for i in range(1,len(s)):
        if s[i] != s[i-1]:
            return i
        i+=1
s = "aabb"
first_non_repeating_char(s)
```
