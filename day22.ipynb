```python
def products(nums):
    answer = []
    product_num = 1
    zero_count = 0
    for num in nums:
        if num == 0:
            zero_count += 1
            continue
        product_num = product_num * num
    for num in nums:
        if zero_count == 1:
            if num == 0:
                answer.append(product_num)
            else:
                answer.append(0)
        elif zero_count > 1:
            return [0]*len(nums)
        else:
            answer.append(int(product_num/num))
    return answer   
nums = [1, 0, 0, 4] 
products(nums)
```

```python
'''
Problem 5:
Given an array of integers, find if there exists a triplet (a, b, c) such that a² + b² = c².
nums = [3, 4, 5]   → True
nums = [3, 4, 6]   → False
nums = [5, 12, 13] → True
'''
```

```python
def triplet(nums):
    counter_square = 0
    for num in nums:
        counter_square += (num**2)
    for num in nums:
        if 2*(num**2) == counter_square:
            return True
    return False
nums = [5, 12, 13] 
triplet(nums)
```
