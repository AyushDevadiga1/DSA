```python
'''  
Problem 1 — 20 minutes, no hints unless stuck.
You are climbing a staircase with n steps. You can climb 1, 2, or 3 steps at a time. How many distinct ways can you reach the top?
n = 3 → answer = 4    # (1+1+1), (1+2), (2+1), (3)
n = 4 → answer = 7

This means for :
n = 0 , answer = 1    # Not climbing 
n = 1 , answer = 1    # (1)
n = 2 , answer = 2    # (1+1) , (2)
'''
def three_staircase(n):
    dp = [1]*(n+1)
    dp[2] = 2
    for i in range (3,n+1):
        dp[i] = dp [i-1] + dp[i-2] + dp[i-3]

    return dp[n]
three_staircase(n=4)
```

```python
''''  
Problem 2 — 20 minutes.
Given a triangle array, find the minimum path sum from top to bottom. At each step you can move to an adjacent number in the row below.
triangle = [
     [2],
    [3,4],
   [6,5,7],
  [4,1,8,3]
]
answer = 11   # 2 + 3 + 5 + 1

Here we have to break this into such a reccurence that we somehow are able to scan each row get the min and sum it to next triangle
'''

def triangles_min_path (triangle):
    n = len(triangle)
    dp = triangle[-1][:]

    for i in range(n-2,-1,-1):
        for j in range(len(triangle[i])):
            print(i,j,triangle[i][j],dp)
            dp[j] = triangle[i][j] + min(dp[j], dp[j+1])
    print(i,j,dp)
    return dp[0]

triangle = [
     [2],
    [3,4],
   [6,5,7],
  [4,1,8,3]
]

triangles_min_path(triangle)
```
