```python
''' 
Problem 1 — Longest Increasing Subsequence
Given an array of integers, find the length of the longest strictly increasing subsequence.
nums = [10, 9, 2, 5, 3, 7, 101, 18]
answer = 4    # [2, 3, 7, 101]

nums = [0, 1, 0, 3, 2, 3]
answer = 4    # [0, 1, 2, 3]

nums = [7, 7, 7, 7]
answer = 1    # no increasing subsequence
Note — subsequence means elements don't have to be contiguous. [2, 3, 7, 101] skips elements but maintains order.
'''
```

```python
def increasing_subsequence(nums):
    dp = [1] * len(nums)
    for i in range(len(nums)):
        for j in range(len(nums)):
            if j < i : 
                if nums[j] < nums[i] :
                    dp[i] = max(dp[i], dp[j] + 1)
    return max(dp)
nums = [7, 7, 7, 7]
increasing_subsequence(nums)
```

```python
''' 
Problem 2 — Longest Common Subsequence
Given two strings, find the length of their longest common subsequence.
 s1 = "abcde", s2 = "ace"
answer = 3    # "ace"

s1 = "abc", s2 = "abc"
answer = 3    # "abc"

s1 = "abc", s2 = "def"
answer = 0    # nothing in common
'''
```

```python
def longest_subsequence(s1,s2):
    dp = [[0] * (len(s2)+1) for _ in range(len(s1)+1)]
    for i in range(1, len(s1)+1):
        for j in range(1, len(s2)+1):
            if s1[i-1] == s2[j-1]:
                dp[i][j] = dp[i-1][j-1] + 1
            else:
                dp[i][j] = max(dp[i-1][j], dp[i][j-1])
    return dp[len(s1)][len(s2)]
s1 = "abc"
s2 = "abc"
longest_subsequence(s1,s2)
```
