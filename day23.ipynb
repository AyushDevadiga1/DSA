```python
''' 
Problem 3:
Given a binary tree, return the level order traversal in reverse — bottom level first.
      3
     / \
    9  20
       / \
      15   7

answer = [[15,7],[9,20],[3]]

'''
```

```python
class TreeNode:
    def __init__(self,val,right=None,left=None):
        self.val = val
        self.left = left
        self.right = right

def reverse_bfs(node):
    from collections import deque
    answer = []
    queue = deque([node])
   
    while queue:
        level_size = len(queue)
        level_items = []
        for _ in range(level_size):
            node = queue.popleft()
            level_items.append(node.val)
            if node.left :
                queue.append(node.left)
            if node.right :
                queue.append(node.right)
        answer.append(level_items)
    return answer[::-1]

root = TreeNode(3)
root.left = TreeNode(9)
root.right = TreeNode(20)
root.right.left = TreeNode(15)
root.right.right = TreeNode(7)

reverse_bfs(root)
```

```python
''' 
Problem 4:

Given a string, find the first non-repeating character and return its index. If none return -1.
s = "leetcode"   → 0
s = "loveleetcode" → 2
s = "aabb" → -1
'''
```

```python
def non_repeating(s):
    hash = {}
    for i in s:
        hash[i] = hash.get(i,0) + 1
    for index,val in enumerate(s):
        if hash[val] == 1:
            return index
    return -1

s = 'aabb'
non_repeating(s)
```

```python
'''
Problem 5:
Given an array of integers, find if there exists a triplet (a, b, c) such that a² + b² = c².
nums = [3, 4, 5]   → True
nums = [3, 4, 6]   → False
nums = [5, 12, 13] → True
'''
```

```python
def triplet(nums):
    nums.sort()
    squares = [n*n for n in nums]
    i,j=0,len(nums)-1
    for k in range(len(nums)-1, 1, -1):  
        while i < j:
            total = squares[i] + squares[j]
            if total == squares[k]:
                return True
            elif total < squares[k]:
                i += 1
            else:
                j -= 1
    return False

nums = [3, 4, 5]   
triplet(nums)
```
