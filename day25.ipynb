```python
''' 
Problem 6:

Given a binary tree and a target sum, return True if there exists a root-to-leaf path that sums to target.
      5
     / \
    4   8
   /   / \
  11  13   4
 /  \       \
7    2        1

target = 22 → True   # path 5→4→11→2
target = 100 → False
'''

class TreeNode:
    def __init__(self,val,left=None,right=None):
        self.val = val
        self.left = left
        self.right = right
    
def pathSum(node, target):
    if node is None:
        return False
    
    target-=node.val

    if target == 0 and node.left is None and node.right is None:
        return True
    
    return pathSum(node.left,target) or pathSum(node.right,target)

root = TreeNode(5)

root.left = TreeNode(4)
root.left.left = TreeNode(11)
root.left.left.left = TreeNode(7)
root.left.left.right = TreeNode(2)

root.right = TreeNode(8)
root.right.left = TreeNode(13)
root.right.right = TreeNode(4)
root.right.right.right = TreeNode(1)

target = 22
pathSum(root,target=target)
```

```python
''' 
Problem 7:

Given a sorted array, remove duplicates in-place and return the new length.
nums = [1,1,2]         → 2, nums = [1,2,...]
nums = [0,0,1,1,1,2,2] → 3, nums = [0,1,2,...]
'''

def removeDuplicates(nums):
    i = 0
    for j in range(1,len(nums)):
        if nums[i] != nums[j]:
            i+=1
            nums[i]=nums[j]
    return nums
nums = [0,0,1,1,1,2,2]
removeDuplicates(nums)
```

```python
''' 
Problem 8:

Given a string s and a pattern p containing only lowercase letters, return all start indices of p's anagrams in s.
s = "cbaebabacd", p = "abc"  → [0,6]
s = "abab", p = "ab"         → [0,1,2]
'''
from collections import Counter

def findAnagrams(s, p):
    answer = []
    p_hash = Counter(p)
    window_hash = Counter(s[:len(p)])

    if p_hash == window_hash : 
        answer.append(0)

    for i in range(len(p),len(s)):
        window_hash[s[i]] += 1
        left_chara = s[ i - len(p)]
        window_hash[left_chara]-=1
        if window_hash[left_chara] == 0:
            del window_hash[left_chara]
        if window_hash == p_hash:
            answer.append(i - len(p) + 1)
    return answer
s = "abab"
p = "ab" 

findAnagrams(s,p)
```
