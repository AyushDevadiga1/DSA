```python
'''  
Problem 3 — 0/1 Knapsack
You have a bag with capacity W. You have n items, each with a weight and value. Maximize the total value without exceeding capacity. Each item can only be used once.
weights = [1, 3, 4, 5]
values  = [1, 4, 5, 7]
W = 7
answer = 9    # items with weight 3+4=7, value 4+5=9

weights = [2, 3, 4]
values  = [3, 4, 5]
W = 5
answer = 7    # items with weight 2+3=5, value 3+4=7
'''
```

```python
def knapsack(weights, values, W):
    dp = [[0] * (W+1) for _ in range(len(weights)+1)]
    for i in range(1, len(weights)+1):
        for j in range(W+1):
            if weights[i-1] <= j:
                dp[i][j] = max(dp[i-1][j], dp[i-1][j-weights[i-1]] + values[i-1])
            else:
                dp[i][j] = dp[i-1][j]
    return dp[len(weights)][W]
weights = [2, 3, 4]
values  = [3, 4, 5]
W = 5
knapsack(weights,values,W)
```

```python
''' 
Problem 1 — Longest Increasing Subsequence
Given an array of integers, find the length of the longest strictly increasing subsequence.
nums = [10, 9, 2, 5, 3, 7, 101, 18]
answer = 4    # [2, 3, 7, 101]

nums = [0, 1, 0, 3, 2, 3]
answer = 4    # [0, 1, 2, 3]

nums = [7, 7, 7, 7]
answer = 1    # no increasing subsequence
Note — subsequence means elements don't have to be contiguous. [2, 3, 7, 101] skips elements but maintains order.
'''
```

```python
def increasing_subsequence(nums):
    dp = [1] * len(nums)
    for i in range(len(nums)):
        for j in range(len(nums)):
            if j < i : 
                if nums[j] < nums[i] :
                    print(f'i = {i} \n j = {j} \n  dp[i] = {dp[i]} \n ')
                    dp[i] = max(dp[i], dp[j] + 1)
    return max(dp)
nums = [0, 1, 0, 3, 2, 3]
increasing_subsequence(nums)
```
