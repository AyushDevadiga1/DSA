```python
'''
Problem 1:
Given an array of integers nums sorted in ascending order, find the starting and ending position of a given target value. If target is not found return [-1, -1].
nums = [5, 7, 7, 8, 8, 10], target = 8
answer = [3, 4]

nums = [5, 7, 7, 8, 8, 10], target = 6
answer = [-1, -1]

nums = [1], target = 1
answer = [0, 0]
'''
```

```python
def func(nums,target):

    answer = []

    result = -1
    low,high = 0,len(nums)-1
    while low <= high : 
        mid = (low+high)//2
        if nums[mid] == target:
            result = mid
            high = mid - 1
        elif nums[mid] > target : 
            high = mid - 1
        else:
            low = mid + 1
    answer.append(result)

    result = -1
    low,high = 0,len(nums)-1
    while low <= high : 
        mid = (low+high)//2
        if nums[mid] == target:
            result = mid
            low = mid + 1
        elif nums[mid] > target : 
            high = mid - 1
        else:
            low = mid + 1
    answer.append(result)

    return answer

nums = [1]
target = 1
func(nums,target)
```

```python
''' 
Problem 2:
Given an array nums and a value k, return the maximum average of any contiguous subarray of length exactly k.
nums = [1, 12, -5, -6, 50, 3], k = 4
answer = 12.75    # subarray [12, -5, -6, 50] / 4 = 12.75 (wait, verify this yourself)

nums = [5, 5, 5, 5, 5], k = 2
answer = 5.0
'''
```

```python
def max_average(nums,k):
    
    if len(nums) < k:
        return -1
    
    low,high = 0,k-1
    current_sum = 0
    for i in range(high+1):
        current_sum += nums[i]
    
    max_avg = current_sum / k
    max_avg = max(max_avg,current_sum/k)

    while high!= len(nums)-1:   
        current_sum-=nums[low]
        low+=1
        high+=1
        current_sum+=nums[high]
        max_avg = max(max_avg,current_sum/k)

    return max_avg

nums = [5,5,5,5,5]
k = 2
max_average(nums,k)
```

```python
'''
Problem 3:
Given a binary tree, return True if it is a valid Binary Search Tree.
A BST is valid if:

Left subtree contains only nodes with values less than the node
Right subtree contains only nodes with values greater than the node
Both subtrees are also valid BSTs

      2
     / \
    1   3
answer = True

      5
     / \
    1   4
       / \
      3   6
answer = False   # 4 is in right subtree of 5 but 4 < 5
'''
```

```python
class TreeNode:
    def __init__(self,val,left=None,right=None):
        self.val = val
        self.left = left
        self.right = right
    
def valid_bst(node, min_val=float('-inf'), max_val=float('inf')):
    if node is None:
        return True
    
    if node.val <= min_val or node.val >= max_val:
        return False
    
    return valid_bst(node.left, min_val, node.val) and valid_bst(node.right, node.val, max_val)

root = TreeNode(5)
root.left = TreeNode(1)
root.right = TreeNode(4)
root.right.left = TreeNode(3)
root.right.right = TreeNode(6)

valid_bst(root)
```
