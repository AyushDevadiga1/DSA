```python
'''
Problem 4:
Given a list of intervals, merge all overlapping intervals.
intervals = [[1,3],[2,6],[8,10],[15,18]]
answer = [[1,6],[8,10],[15,18]]

intervals = [[1,4],[4,5]]
answer = [[1,5]]
'''
```

```python
from collections import deque

intervals = [[1,4],[4,5]]

def merge_intervals(intervals):

    answer = []
    intervals.sort(key=lambda x: x[0])

    current_interval = intervals[0]

    for i in range(1,len(intervals)):

        if intervals[i][0] <= current_interval[1]:

            current_interval = [current_interval[0], max(current_interval[1], intervals[i][1])]

        else : 

            answer.append(current_interval)
            current_interval = intervals[i]

    answer.append(current_interval)

    return answer

merge_intervals(intervals)
```

```python
'''
Problem 5:
Given a string s containing just the characters (, ), {, }, [, ], determine if the input string is valid. A string is valid if:

Open brackets are closed by the same type
Open brackets are closed in the correct order

s = "()"        → True
s = "()[]{}"    → True
s = "(]"        → False
s = "([)]"      → False
s = "{[]}"      → True
'''
```

```python
from collections import deque

def paranthesis(s):
    stack = deque()
    hash = {
             ")" : "(",
             "}" : "{",
             "]" : "["
    }
    for i in s :
        if i in hash:
            temp = stack.pop()
            if temp != hash[i]:
                return False
        else:
            stack.append(i)
        
    return True if not stack else False
                
s = "{[]}" 
paranthesis(s)
```
