```python
# Heap is a special data structure which stores the data in ascending order in a queue like manner
# We can push , push ,pushpop from ends in heap
```

```python
import heapq

nums = [3, 2, 1, 5, 6, 4]
k = 2

def k_largest(nums,k):
    
    heap = [] # We will make this a heap

    for num in nums:

        heapq.heappush(heap,num) # We always push the value only if the value is greater than anyu value in the heap

        if len(heap) > k:
            
            heapq.heappop(heap) # We pop the element from the heap because we need it to be of size k 
            # Always we pop the element at from the left end as it would be the smallest element

        
    return heap[0]  # We need to return the k'th larget element so heap[0] because our size is k and the heap stores value in ascendin order
    

k_largest(nums,k)
```

```python
import heapq

nums = [1, 1, 1, 2, 2, 3]
k = 2

# heap can also handle hash values as well 

def natural_extension(nums,k):

    hash = {} # <-- To store frequencies
    heap = []

    # First loop to store the frequency of each elements
    for num in nums : 

        hash[num] = hash.get(num,0) + 1

    # Second loop to push and pop elements based on heap values

    for num,count in hash.items():
        heapq.heappush(heap,(count,num)) # Always send count first cause it is the main parameter here
        if len(heap) > k :
            heapq.heappop(heap)

    return [num for count,num in heap]

natural_extension(nums,k)
```
