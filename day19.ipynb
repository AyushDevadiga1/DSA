```python
''' 
Problem 6:
Given the head of a linked list, return the node where the cycle begins. If there is no cycle return None.
1 → 2 → 3 → 4 → 5 → 3 (cycle back to node with value 3)
answer = node with value 3

1 → 2 → 3
answer = None
'''
```

```python
class Node:
    def __init__(self,val):
        self.val = val
        self.next = None
        
def detectCycle(head):

    visited = set()
    current = head

    while current is not None:
        if current in visited:
            return current.val
        visited.add(current)
        current = current.next
    return None
    
node1 = Node(1)
node2 = Node(2)
node3 = Node(3)
node4 = Node(4)
node5 = Node(5)
node1.next = node2
node2.next = node3
node3.next = node4
node4.next = node5
node5.next = node3 

detectCycle(node1)
```

```python
'''
Problem 7:
Given an integer array nums, return all the triplets [nums[i], nums[j], nums[k]] such that i != j, i != k, j != k, and nums[i] + nums[j] + nums[k] == 0.
nums = [-1, 0, 1, 2, -1, -4]
answer = [[-1, -1, 2], [-1, 0, 1]]

nums = [0, 0, 0]
answer = [[0, 0, 0]]
'''
```

```python
def three_sum(nums):
    answer = []
    nums.sort()
    for i in range(len(nums)):
        if i > 0 and nums[i] == nums[i-1]:
            continue
        j,k = i+1,len(nums)-1
        while j < k:
            count_sum = nums[i] + nums[j] + nums[k]
            if count_sum == 0:
                answer.append([nums[i], nums[j], nums[k]])
                j += 1
                k -= 1
                while j < k and nums[j] == nums[j-1]:
                    j += 1
                while j < k and nums[k] == nums[k-1]:
                    k -= 1
            elif count_sum > 0:
                k -= 1
            else:
                j += 1
    return answer

nums = [0, 0, 0]
three_sum(nums)
```
